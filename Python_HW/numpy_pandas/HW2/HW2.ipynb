## Задание 1

Напишите функцию, которая классифицирует фильмы из материалов занятия по следующим правилам:

- оценка **2 и ниже** — низкий рейтинг;
- оценка **4 и ниже** — средний рейтинг;
- оценка **4.5 и 5** — высокий рейтинг.

Результат классификации запишите в столбец `class`.

````python
import pandas as pd
````

````python
df_movies = pd.read_csv('movies.csv')
df_ratings = pd.read_csv('ratings.csv')
````

````python
df_movies['movieId'].unique()
````

````python
df_rated = df_ratings.groupby('movieId').mean()['rating']
````

````python
df_rated
````

````python
def rating(rating):
    if rating <= 2:
        return 'Low rating'
    if rating <= 4:
        return 'Medium rating'
    return 'High rating'
````

````python
df_movies['class'] = df_rated.apply(rating)
df_movies['class'] = df_movies['class'].fillna('no_rating')
````

````python
df_movies
````

Используйте файл `keywords.csv`.

Нужно написать **гео-классификатор**, который для каждой строки (поискового запроса) определяет географическую принадлежность к одному из регионов. Если поисковый запрос содержит название города региона, то в столбце `region` нужно записать название этого региона. Если поисковый запрос не содержит ни одного города из словаря, то в `region` ставим значение `undefined`.

Правила распределения по регионам:

```python
geo_data = {
    'Центр': ['москва', 'тула', 'ярославль'],
    'Северо-Запад': ['петербург', 'псков', 'мурманск'],
    'Дальний Восток': ['владивосток', 'сахалин', 'хабаровск']
}
```

Результат классификации запишите в отдельный столбец `region`.
  

geo_data = {
    'Центр': ['москва', 'тула', 'ярославль'],
    'Северо-Запад': ['петербург', 'псков', 'мурманск'],
    'Дальний Восток': ['владивосток', 'сахалин', 'хабаровск']
}

````python
def region(string):
    words = string.lower()
    for reg, cities in geo_data.items():
        for city in cities:
            if city in words:
                return reg
    return 'undefined'
````

````python
df_keywords = pd.read_csv('keywords.csv')
df_keywords['region'] = df_keywords['keyword'].apply(region)
````

````python
df_keywords[df_keywords.region != 'undefined']
````

### Задание 3

Есть мнение, что раньше снимали настоящее кино, не то что сейчас. Ваша задача — проверить это утверждение, используя файлы с рейтингами фильмов из прошлого домашнего занятия: файл `movies.csv` и `ratings.csv` из базы. Нужно проверить, верно ли, что с ростом года выпуска фильма его средний рейтинг становится ниже.

Вы не будете затрагивать субъективные факторы выставления этих рейтингов, а пройдётесь по алгоритму:

1. В переменную `years` запишите список из всех годов с 1950 по 2010 года.
2. Напишите функцию `production_year`, которая каждой строке из названия фильма выставляет год выпуска. Не все названия фильмов содержат год выпуска в одинаковом формате, поэтому используйте алгоритм:
    * Для каждой строки пройдите по всем годам списка `years`;
    * Если номер года присутствует в названии фильма, то функция возвращает этот год, как год выпуска;
    * Если ни один из номеров года списка `years` не встретился в названии фильма, то возвращается 1900 год.
3. Запишите год выпуска фильма по алгоритму пункта 2 в новый столбец `year`.
4. Посчитайте средний рейтинг всех фильмов для каждого значения столбца `year` и отсортируйте результат по убыванию рейтинга.

````python
years = list(range(1950, 2011))
````

````python
def production_year(title):
    for year in years:
        if str(year) in title:
            return year
    return 1900
````

````python
df_movies['year'] = df_movies['title'].apply(production_year)
````

````python
df_movies['rating'] = df_rated
````

````python
df_by_year = df_movies.groupby('year')['rating'].mean().reset_index()
df_by_year.sort_values(by='rating', ascending=False)
````
