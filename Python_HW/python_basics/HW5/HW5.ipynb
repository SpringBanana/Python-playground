# Домашнее задание

Переведите содержимое файла **purchase_log.txt*** в словарь `purchases` вида:
{‘1840e0b9d4’: ‘Продукты’, …}

Для каждого `user_id` в файле **visit_log.csv*** определите третий столбец с категорией покупки, если покупка была. Сам файл **visit_log.csv*** изменять не надо.

Запишите в файл **funnel.csv** визиты из файла **visit_log.csv***, в которых были покупки с указанием категории.

## Учтите условия на данные:
- содержимое **purchase_log.txt*** помещается в оперативную память компьютера;
- содержимое **visit_log.csv*** — нет; используйте только **построчную обработку** этого файла.

```python
import json

purchases = {}
with open('purchase_log.txt', 'r', encoding='utf-8') as f_purchases:
    for line in f_purchases:
        data = json.loads(line.strip())
        user_id = data['user_id']
        purchases.setdefault(user_id, []).append(data['category'])

with open('visit_log.csv', 'r', encoding='utf-8') as f_visits, \
     open('funnel.csv', 'w', encoding='utf-8') as f_results:
    for line in f_visits:
        data = line.strip().split(',')
        user_id, source = data[0], data[1]
        if user_id in purchases:
            categories = ','.join(purchases[user_id])
            f_results.write(f"{user_id},{source},{categories}\n")
        else:
            f_results.write(f"{user_id},{source},\n")
```
