```python
books = [
    {'genre': 'поэзия', 'number': '978-5-1000-1234-7', 'title': 'Евгений Онегин', 'author': 'Александр Пушкин'},
    {'genre': 'фэнтези', 'number': '88006', 'title': 'Властелин колец', 'author': 'Джон Р. Р. Толкин'},
    {'genre': 'детектив', 'number': 'D-1122', 'title': 'Безмолвный свидетель', 'author': 'Агата Кристи'}
]
directories = {
    '1': ['978-5-1000-1234-7', '88006'],
    '2': ['D-1122'],
    '3': []
}

global books, directories
```

```python
def book_info(book_id):
    for book in books:
        if book['number'] == book_id:
            return book['title'], book['author']
    return
```

```python
def shelf(title):
    book_id = None
    for book in books:
        if book['title'] == title:
            book_id = book['number']
            break
    if book_id is None:
        return
    for key, value in directories.items():
        if book_id in value:
            return key
```

```python
def all():
    for book in books:
        print(f"№: {book['number']}, жанр: {book['genre']}, название: {book['title']}, автор: {book['author']}, полка хранения: {shelf(book['title'])}")
```

```python
def add_shelf(shelf):
    if shelf in directories.keys():
        print(f"Такая полка уже существует. Текущий перечень полок: {', '.join(list(directories.keys()))}")
    else:
        directories.setdefault(shelf, [])
        print(f"Полка добавлена. Текущий перечень полок: {', '.join(list(directories.keys()))}")
```

```python
def del_shelf(shelf):
    if shelf in directories.keys():
        if not directories[shelf]:
            del directories[shelf]
            print(f"Полка удалена. Текущий перечень полок: {', '.join(list(directories.keys()))}")
        else:
            print(f"На полке есть книги, удалите их перед удалением полки. Текущий перечень полок: {', '.join(list(directories.keys()))}")
    else:
        print(f"Такой полки не существует. Текущий перечень полок: {', '.join(list(directories.keys()))}")
```

```python
def add_book():
    book_id = input('Введите номер книги: ')
    genre = input('Введите жанр книги: ')
    title = input('Введите название книги: ')
    author = input('Введите автора книги: ')
    shelf = input('Введите полку для хранения: ')
    if (book_info(book_id) is not None):
        print("Книга с таким номером уже существует")
    if (shelf not in directories):
        print("Такой полки не существует. Добавьте полку командой add_shelf.")
    if ((book_info(book_id) is None) and (shelf in directories)):
        new_book = {
            'genre': genre,
            'number': book_id,
            'title': title,
            'author': author}
        directories[shelf].append(book_id)
        books.append(new_book)
        print("Книга добавлена ", end="")
    print("Текущий список книг:")
    all()
```

```python
def del_book():
    book_id = input('Введите номер книги: ')
    for book in books[:]:
        if book['number'] == book_id:
            directories[shelf(book_info(book_id)[0])].remove(book_id)
            books.remove(book)
            print("Книга удалена.")
            print("Текущий список книг:")
            all()
            return
    print("Книга не найдена в базе.")
    print("Текущий список книг:")
    all()
```

```python
def move():
    book_id = input('Введите номер книги: ')
    target_shelf = input('Введите номер полки: ')
    if book_info(book_id) is None:
        print("Книга не найдена в базе.")
        print("Текущий список книг:")
        all()
        return
    if target_shelf not in directories:
        print(f"Такой полки не существует. Текущий перечень полок: {', '.join(list(directories.keys()))}")
        return
    if (book_id in directories[target_shelf]):
        print("Книга уже на этой полке.")
        print("Текущий список книг:")
        all()
        return
    if ((book_info(book_id) is not None) and (target_shelf in directories)):
        directories[shelf(book_info(book_id)[0])].remove(book_id)
        directories[target_shelf].append(book_id)
        print("Книга перемещена.")
        print("Текущий список книг:")
        all()
```

```python
def main():
    cmd = None
    while cmd != 'q':
        cmd = input("Введите команду:")
        match cmd:
            case 'book_info':
                book_id = input("Введите номер книги:")
                res = book_info(book_id)
                print(("Книга не найдена в базе" if res is None 
                   else f"Название книги: {res[0]}\nАвтор: {res[1]}"))
            case 'shelf':
                title = input("Введите название книги:")
                res = shelf(title)
                print(("Книга не найдена в базе" if res is None 
                   else f"Книга хранится на полке: {res}"))
            case 'all' :
                all()
            case 'add_shelf':
                add_shelf(input("Введите номер полки:"))
            case 'del_shelf':
                del_shelf(input("Введите номер полки:"))
            case 'add_book':
                add_book()
            case 'del_book':
                del_book()
            case 'move':
                move()
            case 'q':
                print("До новых встреч")
            case _:
                print("Не верная команда попробуйте еще")
```

```python
main()
```
