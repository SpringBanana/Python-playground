### Задание 1

Напишите функцию, которая возвращает **название валюты** (поле `Name`) с **максимальным значением курса** с помощью сервиса [https://www.cbr-xml-daily.ru/daily_json.js](https://www.cbr-xml-daily.ru/daily_json.js).

```python
from exchange import Rate
import requests
```

```python
class CurrencyName(Rate):
    """
    Класс для получения названия валюты с максимальным курсом к рублю
    из текущих данных ЦБ РФ.

    Работает как расширение класса Rate: использует его логику получения
    данных об обменных курсах, но дополнительно реализует поиск валюты
    с максимальным курсом.
    """
    def __init__(self, format_ = 'value'):
        super().__init__(format_)

    def max_currency(self):
        """
        Возвращает название валюты с максимальным курсом к рублю.

        Returns
        -------
        str
            Название валюты с максимальным значением 'Value' среди всех валют
            в текущих данных ЦБ РФ.
        """
        valutes = self.exchange_rates()
        max_currency = max(
        valutes.items(),
        key=lambda x: x[1]['Value']
        )
        return max_currency[1]['Name']
```
